```python
result = app_graph.invoke(
            {
                "query": "what is leave policy?",
                "chat_history": []
            },
        )
print("RAG Result:", result["answer"])
context = "\n\n".join(
    doc.page_content
    for doc in result["documents"]
)
faithfulness_score = evaluate_faithfulness(
    judge_llm,
    question=result["query"],
    context=context,
    answer=result["answer"]
)

answer_relevance_score = evaluate_answer_relevance(
    judge_llm,
    question=result["query"],
    answer=result["answer"]
)

print("Faithfulness:", faithfulness_score)
print("Answer Relevance:", answer_relevance_score)  
```
